Day 5: Exploratory Data Analysis

```python
import pandas as pd
df = pd.read_csv("sales_cleaned.csv")
top_3 = df.groupby('Product')['Sales'].sum().nlargest(3)
print(top_3)
```

```python
import matplotlib.pyplot as plt

top_3.plot(kind='bar')

plt.title("Top 3 Performing Products")
plt.xlabel("Product")
plt.ylabel("Total Revenue")

plt.show()
```

Reflection: Correlation vs Causation

A high correlation (like 0.85) between Product A and Product B does not mean that one causes the other to sell. Correlation only indicates that the two variables move together, not that one directly influences the other.

There could be other factors affecting both products, such as seasonal demand, marketing campaigns, or customer preferences. Therefore, correlation does not imply causation.
