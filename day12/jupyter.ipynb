Day 12: Business Dashboarding

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/HP/Downloads/data_analyst/day10/sales_cleaned.csv")

df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Sales'] = df['Price'] * df['Quantity']

region_df = df[df['Region'] == 'North']

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('North Region Business Dashboard', fontsize=20)

sns.lineplot(ax=axes[0, 0], x='Date', y='Sales', data=region_df)
axes[0, 0].set_title('North Region Sales Trend')

sns.barplot(ax=axes[0, 1], x='Category', y='Sales', data=region_df, estimator=sum)
axes[0, 1].set_title('Revenue by Category (North)')

region_df['Category'].value_counts().plot.pie(
    ax=axes[1, 0],
    autopct='%1.1f%%'
)
axes[1, 0].set_title('Category Distribution (North)')

sns.scatterplot(ax=axes[1, 1], x='Price', y='Sales', data=region_df)
axes[1, 1].set_title('Price vs Sales (North)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

Reflection

A 4-chart dashboard is more effective than sending separate image files because it provides a unified and structured view of key business metrics in one place. This allows managers to quickly compare trends, relationships, and performance without switching between multiple files.

Dashboards improve decision-making by presenting data in context. For example, a manager can simultaneously observe sales trends, category performance, and pricing relationships, which helps in identifying patterns and making faster, more informed decisions.

In contrast, separate images break the flow of analysis, increase cognitive load, and make it harder to connect insights across different charts. Therefore, dashboards enhance clarity, efficiency, and overall business understanding.
