Day 18: Recommendation Logic through Association

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

data = {
    'TransactionID': [1,1,1,2,2,3,3,3,4,4,5,5],
    'Activity': [
        'Coding','Chess','Pizza',
        'Coding','Chess',
        'Hiking','Photography','Pizza',
        'Hiking','Photography',
        'Coding','Pizza'
    ]
}

df = pd.DataFrame(data)

basket = (
    df.groupby(['TransactionID', 'Activity'])['Activity']
    .count()
    .unstack()
    .fillna(0)
)

basket_sets = (basket > 0).astype(int)

frequent_itemsets = apriori(basket_sets, min_support=0.2, use_colnames=True)

rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

rules = rules.sort_values(by='confidence', ascending=False)

filtered_rules = rules[
    (rules['lift'] > 1.5) & (rules['confidence'] > 0.6)
]

top_3 = filtered_rules.sort_values(by='lift', ascending=False).head(3)

print("Top Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])

print("\nTop 3 Premium Bundles:")
print(top_3[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

Reflection:
No, this is not a reliable rule for a marketing campaign. Even if the association between Yoga and Juice Tasting shows high confidence or lift, a support of only 1% means that this pattern occurs in a very small portion of the data. This makes it statistically weak and not scalable for business decisions. Relying on such a rare pattern could lead to ineffective campaigns because it does not represent the behavior of the majority of users. For marketing, we should prioritize rules that are not only strong (high lift and confidence) but also sufficiently frequent (higher support) to ensure broader impact.
