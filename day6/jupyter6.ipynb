```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/HP/Downloads/data_analyst/day6/sales_data_with_product.csv")

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='Price',
    y='Quantity',
    hue='Product'
)

plt.title("Price vs Quantity by Product")
plt.xlabel("Price")
plt.ylabel("Quantity")

plt.tight_layout()
plt.show()
```

Reflection

A Box Plot helps identify data errors by showing the full distribution of data, including outliers. Extreme values appear as individual points outside the whiskers, making them easy to detect.

In contrast, a Bar Chart showing averages can hide these errors because it compresses all values into a single number. For example, a very high or incorrect sales value may significantly affect the average, but the bar chart will not reveal that it is an outlier.

Thus, a Box Plot provides better visibility into anomalies and data quality issues compared to a simple average-based Bar Chart.
