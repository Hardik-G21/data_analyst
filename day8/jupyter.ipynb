Day 8: Business Scenario & Aggregation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("C:/Users/HP/Downloads/data_analyst/day8/sales_cleaned.csv")

df['Sales'] = df['Price'] * df['Quantity']

df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')

filtered_df = df[df['Product'] == 'Product A']

filtered_df = filtered_df.sort_values('Date')

filtered_df['Cumulative_Sales'] = filtered_df['Sales'].cumsum()

plt.figure(figsize=(8, 5))

sns.lineplot(
    data=filtered_df,
    x='Date',
    y='Cumulative_Sales',
    marker='o'
)

plt.title("Cumulative Sales Trend - Product A")
plt.xlabel("Date")
plt.ylabel("Cumulative Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Reflection

It is important to analyze 'Transaction Count' alongside 'Total Revenue' because revenue alone does not provide the full picture of business performance.

A high revenue number can sometimes be misleading if it is driven by only a few large transactions rather than consistent sales. For example, a product may show high total revenue due to one or two bulk purchases, but have very low transaction frequency overall.

By comparing transaction count, analysts can understand whether revenue is stable and repeatable or dependent on rare events. This helps in making better decisions related to customer behavior, product demand, and long-term business sustainability.
