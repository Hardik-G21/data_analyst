Day 16: A/B Testing & Significance

```python
import numpy as np
from scipy import stats

np.random.seed(42)

# Small Sample (5 days)
group_a_small = np.random.normal(50, 10, 5)
group_b_small = np.random.normal(58, 12, 5)

t_small, p_small = stats.ttest_ind(group_a_small, group_b_small)

print("=== SMALL SAMPLE (5 Days) ===")
print(f"P-Value: {p_small:.4f}")

# Large Sample (500 days)
group_a_large = np.random.normal(50, 10, 500)
group_b_large = np.random.normal(58, 12, 500)

t_large, p_large = stats.ttest_ind(group_a_large, group_b_large)

print("\n=== LARGE SAMPLE (500 Days) ===")
print(f"P-Value: {p_large:.4f}")
```
