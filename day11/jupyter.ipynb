Day 11: Correlation & Feature Relationships

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("C:/Users/HP/Downloads/data_analyst/day10/sales_cleaned.csv")

# Create Sales column
df['Sales'] = df['Price'] * df['Quantity']

# -------------------------------
# Step 1: Correlation Matrix
# -------------------------------
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_df.corr()

print("Correlation Matrix:")
print(corr_matrix)

# -------------------------------
# Step 2: Find strongest correlation with Sales
# -------------------------------
sales_corr = corr_matrix['Sales'].drop('Sales')
top_feature = sales_corr.idxmax()

print(f"Highest correlated feature with Sales: {top_feature}")

# -------------------------------
# Step 3: Scatter Plot
# -------------------------------
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df[top_feature], y=df['Sales'])

plt.title(f"{top_feature} vs Sales (Scatter Plot)")
plt.xlabel(top_feature)
plt.ylabel("Sales")

plt.show()

# -------------------------------
# Step 4: Regression Plot
# -------------------------------
plt.figure(figsize=(8, 5))
sns.regplot(x=df[top_feature], y=df['Sales'], scatter_kws={'alpha':0.6})

plt.title(f"{top_feature} vs Sales (Regression Line)")
plt.xlabel(top_feature)
plt.ylabel("Sales")

plt.show()
```

Reflection

A high correlation (such as 0.85 between Discount and Sales) does not necessarily mean that discounts caused the increase in sales. Correlation only indicates that two variables move together, not that one directly causes the other.

There could be other factors influencing both variables. For example, higher discounts might be offered during peak seasons, promotions, or special events, which themselves drive higher sales. This means the relationship may be influenced by external factors rather than direct causation.

This concept is known as "Correlation vs. Causation." Just because two variables are strongly related does not mean one is responsible for the change in the other. Analysts must investigate further, often using experiments or controlled studies, to establish true causation.
