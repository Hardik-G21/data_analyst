Day 15: Time-Series & Trend Smoothing

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "daily_registrations.csv",
    parse_dates=['Date'],
    index_col='Date',
    dayfirst=True
)

df['MA_3'] = df['Registrations'].rolling(window=3).mean()
df['MA_30'] = df['Registrations'].rolling(window=30).mean()

plt.figure(figsize=(12,6))

plt.plot(df.index, df['Registrations'], alpha=0.3, label='Actual')
plt.plot(df.index, df['MA_3'], label='3-Day MA')
plt.plot(df.index, df['MA_30'], label='30-Day MA')

plt.title("3-Day vs 30-Day Moving Average")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

Reflection:
A 30-day moving average is more useful for planning a major marketing push because it reflects the underlying long-term trend rather than short-term fluctuations. Data from just the last 3 days may include noise, sudden spikes, or drops that are not representative of typical user behavior. The 30-day average smooths out these irregularities and provides a more stable and reliable estimate of expected performance, helping in better decision-making and resource allocation.
