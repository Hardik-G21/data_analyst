```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
users = pd.read_csv("users.csv")
sales = pd.read_csv("sales.csv")

merged = pd.merge(users, sales, on="ID", how="inner")

print("Merged Data:\n")
print(merged)
```

```python
import pandas as pd

df = pd.read_csv("C:/Users/HP/Downloads/data_analyst/sample_data.csv", sep="\t")

city_stats = df.groupby("City")["Salary"].mean()
print("Average Salary by City:\n")
print(city_stats)


pivot = df.pivot_table(
    values="Salary",
    index="City",
    columns="Age",
    aggfunc="mean"
)

print("\nData Pivot Table:\n")
print(pivot)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/HP/Downloads/data_analyst/sample_data.csv", sep="\t")

df.groupby("City")["Salary"].mean().plot(kind='bar', color='skyblue')

plt.title("Average Salary by City")
plt.ylabel("Salary")
plt.show()

df["Salary"].plot(kind='hist', bins=5, edgecolor='black')

plt.title("Salary Distribution")
plt.show()
```

Merging data from two different files allows a business to combine related information into a single, meaningful dataset. For example, user details and sales data can be joined to understand which customers are generating revenue. This helps in identifying high-value customers, analyzing regional performance, and making data-driven decisions such as targeted marketing and resource allocation.
